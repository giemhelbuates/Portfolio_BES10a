## Program 1 - Password Strength Checker

#### Create a function that checks if a password is weak, average, or strog based on its length

##### Function to check password strength

```python
def check_password(password):
    if len(password) <=4:
        return "Weak Password" #Returned if password is too short
    elif len(password) <=8:
        return "Average Password" #Returned if password is moderate
    else:
        return "Strong Password" #Returned if password is long
```

##### Prompt the user to enter a password

```python
password = input("Enter your password: ") #Stores the password
```

```python
print ("Password:", password) #Displays entered password
```

##### Function call

```python
result = check_password(password) #Calls the function and stores the result
```

##### Display the result 

```python
print ("Password Strength: ", result) #Displays password strength
```

## Program 2 - Secret Message Encoder

```python
#Function to encode the messae
def encode_message(text):
    result ="" #Stores the encoded message
    for char in text:
        result = result + char*2 #Repeats every character two times
    return result #Return the encoded message

#Prompt the user to enter a message
message = input("Enter a word or message: ") #Stores the message

print ("message:", message) #Displays entered message

#Call the Function
encoded = encode_message(message) #Calls the function and stores the result

#Display the resut
print("Encoded Message:", encoded) #Displays encoded message
```

## Program 3 - Student Contribution Calculator

##### Function to compute each student's contribution

```python
def compute_contribution(total_cost, students):
    contribution = total_cost / students #Computes contribution per student
    return contribution #Returns the computed contribution
```

##### Prompt the user to enter project details

```python
total_cost = float(input("Enter the total project cost: ")) #Stores total project cost
students = int(input("Enter the number of students: ")) #Stores the number of students
```

```python
print ("Total Project total project cost:", total_cost) #Displays entered total project cost
print ("Total Number of Students:", students) #Displays entered number of students
```

##### Call the Function and Display the Result

```python
payment = compute_contribution(total_cost,students) #Calls the function and stores the result
print ("Each Student Should Contribute: ", payment) #Displays contribution per student
```

```python
print ("Program is Finished!")
```

#### This program uses a function to compute how much each student should contribute for a class project.
