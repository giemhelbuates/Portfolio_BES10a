### Program 1: Suspicious WiFi Name Checker

##### Create empty List

```python
wifi_names = list() #Stores WiFi names
suspicious_wifi = list() #Stores suspicious WiFi names
```

##### Prompt the user to enter number of WiFi names

```python
count = int(input("Enter number of WiFI names: ")) #Stores number of WiFI names
```

```python
for number in range(count):
    wifi = input("Enter WiFi name; ")  #User enters WiFi name
    wifi_names.append(wifi) #Adds WiFi name into list

    # convert to lowercase
    check_wifi = wifi.lower()

    # check suspicious words
    if "hack" in check_wifi:
        suspicious_wifi.append(wifi) #Adds suspicious WiFi name

    elif "virus" in check_wifi:
        suspicious_wifi.append(wifi) #Adds suspicious WiFi name

    elif "freewifi" in check_wifi:
        suspicious_wifi.append(wifi) #Adds suspicious WiFi name
```

##### Display Results

```python
print ("All WiFi Names:", wifi_names) #Display all WiFi names
print ("Suspicious WiFi names:" , suspicious_wifi) #Displays suspicious WiFi names

#Checks if suspicious WiFi names exist
if len (suspicious_wifi) == 0:
    print ("No Suspicious WiFi Name Found") #Displayed if none found
else: 
    print ("Be Careful with These WiFi Networks") #Displayed if suspicious names exist
```

### Program 2: Group Project Contribution Tracker

```python
# Create empty Lists
members = list () #Stores group member
completed = list () #Stores members with completed tasks

# Prompt the user to enter number of group members
count = int(input("Enter number of group members: ")) #Stores number of members

#Loop for group members
for number in range (count):
    name = input("Enter group member name: ") #Stores member name
    status = input("Dis the member finish the task? (yes/no): ") #Stores completion status

    members.append(name) #Add member name into list

#Check completion status
if status.lower()== "yes":
    completed.append(name) #Adds completed into list

print ("Group Members: ", members) #Display all group members
print ("Members Who Finished Tasks:", completed) #Displays completed members

# Computes unfinished members
unfinished = len(members) - len (completed) #Computes unfinished members

print ("Number of Mmembers Without Finished Tasks:", unfinished) # Displays unfinished members
```
