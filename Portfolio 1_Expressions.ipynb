# Program 2: Student Average Grade 

#### Write a program to prompt the user to input grades and calculate the average grade of a student based on major subjects for the first semester

##### Prompt the user to input grades for major subjects

```python
math10 = float(input("Enter grade for Math 10: "))
che110 = float(input("Enter grade for ChE 110: "))
chem11 = float(input("Enter grade for Chem 11: "))
chem11L = float(input("Enter grade for Chem 11L: "))
math11 = float(input("Enter grade for Math 11: "))
```

##### Compute Average Grade 

```python
average = (math10+ che110 + chem11 + chem11L + math11)/5
```

##### Display Results

```python
print("\n----- STUDENT AVERAGE GRADE IN MAJOR SUBJECTS -----")
print("Math 10:", math10)
print("ChE 110:", che110)
print("Chem 11:", chem11)
print("Chem 11L:", chem11L)
print("Math 11:", math11)

print("\nAverage Grade:", average)
```

#### This program demonstrates how variables and expressions are used to compute the average grade of a student based on major subjects. It show how input data can be processed using arithmetic operations to produce meaningful results by allowing the program to automatically compute values instead of doing manual calculations, making it more efficient and accurate.

# Program 3: My School Transportation Expenses

#### Write a program that prompts the user to compute daily and weekly transportation based on school travel fares.

##### FARE PER ONE-WAY TRIP

###### Fare from home riding a jeepney to Legazpi

```python
fare_legazpi = 38
```

###### Fare riding a jeepney to Albay

```python
fare_albay = 13
```

###### Fare riding a tricycle to BUCENG

```python
fare_buceng = 10
```

###### Number of School days

```python
school_days = 6
```

##### COMPUTE ONE-WAY TRIP FARE (home → school)

```python
one_way_fare = fare_legazpi + fare_albay + fare_buceng
```

##### ROUND TRIP FARE (to school + going home)

```python
daily_fare = one_way_fare * 2
```

##### WEEKLY TRANSPORTATION EXPENSE

```python
weekly_expense = daily_fare * school_days
```

##### Display Results

```python
print("\n----- TRANSPORTATION EXPENSE SUMMARY -----")
print("One-way Fare:", one_way_fare)
print("Daily Fare (Round Trip):", daily_fare)
print("Weekly Transportation Expense:", weekly_expense)
```

#### *This program helps me estimate my daily and weekly transportation expenses for school.*
