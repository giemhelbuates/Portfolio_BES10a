### Focus Study Loop Tracker

#### An iteration program that tracks study sessions per subject.

##### Prompt the user for number of subjects studied

```python
subjects = int(input("Enter number of subjects studied: ")) #Stores number of subjects
print ("Enter number of subjects studied: ", subjects) #Displays subjects studied
```

```python
# Initial Values
total_hours = 0 #Stores total study hours
session_count = 0  #Stores number of extra study sessions

#List of subjects
subject_list = ["Calculus", "Physics", "Chemistry", "Programming"]
```

##### For Loop Section (definite iteration)

```python
for number in range (subjects):
    hours = float(input("Enter study hours: ")) #Stores study hours input
    total_hours += hours #Adds input to total study hours
    print ("Enter study hours", hours)

    print("Study time recorded for subject", number + 1) #Displays progress per loop
```

##### While Loops Section (indefinite iteration)

```python
print("\n===== EXTRA STUDY SESSION LOGGER =====") #Displays section title
print ("Type 'stop' to end logging") 
while True:
    session = input("Enter extra study session: ") #Stores session input
    print ("Enter extra study session", session)
    if session.lower () == "stop":
        break #Stops the loop when user types stop
    session_count += 1 #Counts valid sessions
    print ("Session",session_count, "recorded successfully") #Confirmation message
```

##### Display Results

```python
print ("\n===== RESULTS =====") #Displays results title
print ("Total Study Hours: ", total_hours) #Displays total hours
print ("Total Extra Sessions:", session_count) #Displays session count
```

##### Conditional Statment (decision making)

```python
if total_hours>=8:
    print ("Excellent! You had a productive study day.") #High performance message
else:
    print ("Keep going! Try to increase your study hours") #Encouragement message
```
